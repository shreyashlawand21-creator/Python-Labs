```python
def seat_countdown(seat):
    if seat == 0:
        print("All seats are booked")
        return
    print("Remaining Seats  :", seat)
    seat_countdown(seat - 1)
```

```python
seat_countdown(5)
```

```python
def final_booking(train_name = "Rajdhani Express", , * seat_number)
    print("-------------------------------------")
    print("-----Indian Railway Reservation------")
    print("-------------------------------------")
    print("Train Name  :",train_name)
    for seat in seat_number:
        print("Allocated Seat Numbers:", seat)
```
